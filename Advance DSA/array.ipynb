```python
# Write a brute force code
ar = [2,5,11,13,7]
k = 9
for i in range(len(ar)):
    for j in range(i + 1, len(ar)):
        if ar[i] + ar[j] == k:
            print("The combination for the sum of k=9 is",ar[i],"and", ar[j])
```

```python
# 3 kai liye hai yai
ar = [2,5,11,13,7,1,4,10,11,43,20,14,17]
k = 23
for i in range(len(ar)):
    for j in range(i + 1, len(ar)):
        for m in range(j+1, len(ar)):
            if ar[i] + ar[j]+ar[m] == k:
                print("The combination for the sum of k=23 is",ar[i],"and", ar[j],"and",ar[m])
```

```python
#pointer kai liye hai yai 
ar = [2, 5, 11, 13, 7]
k = 9
ar.sort()
left = 0
right = len(ar) - 1
while left < right:
    current_sum = ar[left] + ar[right]
    if current_sum == k:
        print("The combination for the sum of k=9 is",
                ar[left], "and", ar[right])
        break
    elif current_sum < k:
        left += 1
    else:
        right -= 1
```

```python
#array se max number find karna hai har tareeke ko use kar 
arr = [10, 25, 7, 45, 32, 18]
maxi = max(arr)
print("max number=",maxi)
```

```python
#loop wala tareeka
arr = [10, 25, 7, 45, 32, 18]
maxi = arr[0]
for num in arr:
    if num > maxi:
        maxi = num
print("max number=",maxi)
```

```python
#sorting wala tareeka 
arr = [10, 25, 7, 45, 32, 18]
arr.sort()
maxi = arr[-1]
print("max number=",maxi)
```

```python
# 3 test-case banao function kai liye taki pata kar sake ki number prime hai ya nahi 
def prime(n):
    if not isinstance(n, int):
        print("the given input is not in the integer form")
        return
    if n < 2:
        print("the number is below 2 which doesn't come in prime")
        return
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            print("the number is not prime")
            return
    print("the number is prime")
prime(7)
prime(10)
prime(1)
prime("A")
prime("-")
prime('-')

```
