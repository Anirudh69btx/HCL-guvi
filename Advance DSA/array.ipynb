```python
# Write a brute force code
ar = [2,5,11,13,7]
k = 9
for i in range(len(ar)):
    for j in range(i + 1, len(ar)):
        if ar[i] + ar[j] == k:
            print("The combination for the sum of k=9 is",ar[i],"and", ar[j])
```

```python
#pointer kai liye hai yai 
ar = [2, 5, 11, 13, 7]
k = 9
ar.sort()
left = 0
right = len(ar) - 1
while left < right:
    current_sum = ar[left] + ar[right]
    if current_sum == k:
        print("The combination for the sum of k=9 is",
                ar[left], "and", ar[right])
        break
    elif current_sum < k:
        left += 1
    else:
        right -= 1
```

```python
# 3 test-case banao function kai liye taki pata kar sake ki number prime hai ya nahi 
def prime(n):
    if not isinstance(n, int):
        print("the given input is not in the integer form")
        return
    if n < 2:
        print("the number is below 2 which doesn't come in prime")
        return
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            print("the number is not prime")
            return
    print("the number is prime")
prime(7)
prime(10)
prime(1)
prime("A")
prime("-")
prime('-')

```

```python
#Find the frequency of each element in an array using a dictionary.
arr=[10, 12, 22, 13, 10, 14, 12, 13]
freq ={}
for element in arr:
    if element in freq:
        freq[element] += 1
    else:
        freq[element] = 1
print(freq)
```

```python
#Check whether a string is a palindrome using Tivo Pointer
def pali(sstr):
    l, r= 0,len(sstr) - 1
    while l<r:
        if sstr[l]!=sstr[r]:
            print("not palindrome")
            return
        l+=1
        r-=1
    print("palindrome")
    return
string = "munmnum"
st="cutietuc"
print(pali(string)) 
print(pali(st)) 
```
