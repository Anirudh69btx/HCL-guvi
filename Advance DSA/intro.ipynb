```python
#student result system-details, marks,grade, percentage,display the result
student = []

def Stud_result():
    stud_name = input("Enter the name of the student = ")
    stud_class_and_section = input("Enter the class and section of student (eg: 4A) = ")
    stud_roll_number = int(input("Enter the roll number of the student = "))
    stud_clg_id = input("Enter the student ID = ")
    stud_age = int(input("Enter the age of the student = "))
    stud_mother_name = input("Enter student's mother name = ")
    stud_father_name = input("Enter student's father name = ")
    total_stud_marks = float(input("Enter the marks obtained by the student = "))
    total_marks_subject = int(input("Enter the total marks = "))
    student_address = input("Enter the address = ")
    stud_per = (total_stud_marks / total_marks_subject) * 100
    if stud_per >= 90:
        grade = "A"
    elif stud_per >= 80:
        grade = "B"
    elif stud_per >= 70:
        grade = "C"
    elif stud_per >= 60:
        grade = "D"
    elif stud_per >= 50:
        grade = "E"
    else:
        grade = "REPEAT_CLASS"
    student.append(stud_name)
    student.append(stud_clg_id)
    student.append(stud_age)
    student.append(stud_roll_number)
    student.append(student_address)
    student.append(stud_class_and_section)
    student.append(stud_mother_name)
    student.append(stud_father_name)
    student.append(total_stud_marks)
    student.append(total_marks_subject)
    student.append(stud_per)
    student.append(grade)
    print("\n----- STUDENT RESULT -----")
    print("Name:", stud_name)
    print("Class & Section:", stud_class_and_section)
    print("Roll Number:", stud_roll_number)
    print("Student ID:", stud_clg_id)
    print("Age:", stud_age)
    print("Mother Name:", stud_mother_name)
    print("Father Name:", stud_father_name)
    print("Address:", student_address)
    print("Marks:", total_stud_marks, "/", total_marks_subject)
    print("Percentage:", stud_per, "%")
    print("Grade:", grade)


Stud_result()
```
