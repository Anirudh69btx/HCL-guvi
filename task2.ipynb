```python
#Mart_Store 
print("Hello!!Welcome to SMS Mart")
GST = 18 
categories = {

"Veggies": [ {"name": "SPINACH", "amount": 32.00, "discount": 2}, {"name":"Mango", "amount":120, "discount":10}],
"Beauty": [ {"name": "Maskara", "amount": 320, "discount": 35}, {"name":"Lipstick", "amount":750, "discount":25}],
"Electronics": [ {"name": "GPU", "amount": 5600, "discount": 275}, {"name":"RGB", "amount":3575, "discount":75}],
"Grocery": [ {"name": "Pulses", "amount": 200, "discount": 15}, {"name":"Kidney Beans", "amount":375, "discount":5}],
"Stationary": [ {"name": "Copy", "amount": 100, "discount": 5}, {"name":"Clay", "amount":75, "discount":2} ],
"Furniture": [ {"name": "Chair", "amount": 15600, "discount": 2115}, {"name":"Table", "amount":13575, "discount":1725}] 

} 
for category, products in categories.items():
    print(f"category={category}")

for product in products:
    amount = product["amount"] 
    discount = product["discount"] 
    discount_price = amount - discount 
    gst_amount = discount_price * GST / 100 
    actual_amount = discount_price + gst_amount 


    print(f"Product: {product['name']}") 
    print(f"Original Amount: ₹{amount:}") 
    print(f"Discount: ₹{discount:}") 
    print(f"Discounted Price: ₹{discount_price:}") 
    print(f"GST: ₹{gst_amount:}") 
    print(f"Final Amount: ₹{actual_amount:}")



```
